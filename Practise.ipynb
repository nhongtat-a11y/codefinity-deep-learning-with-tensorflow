```python
import torch
```

```python
# create 6 x 8 tensor with random values between 0 and 1
```

```python
random_tensor = torch.rand(6, 8)
print(random_tensor)
```

```python
normal_tensor = torch.randn(6,8)
print(normal_tensor)
```

```python
# create 4x3 tensor with random integer between 0 and 10
integer_tensor = torch.randint(0, 10, (4,3))
print(integer_tensor)
```
