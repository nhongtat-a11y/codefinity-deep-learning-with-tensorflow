# Assignment 5.1-Tokenizer
A tokenizer is a tool or process used in natural language processing (NLP) that analyzes text to break down text into smaller, manageable units called tokens. These tokens can be words, phrases, symbols, or other meaningful elements. The primary purpose of tokenization is to prepare text for further processing, such as parsing, analysis, or modeling.

There are different types of Tokenizers


1.   Word Tokenizers
2.   Subword Tokenizers
3.   Character Tokenizers
4.   Sentence Tokenizers
5.   Whitespace Tokenizers
6.   Punctuation-Based Tokenizers
7.   Rule-Based Tokenizers
8.   Pretrained Tokenizers
9.   Hybrid Tokenizers

**Word Tokenizers**: These tokenizers split text into individual words.

**Sub-Word Tokenizers**: These tokenizers break down words into smaller units, such as prefixes, suffixes, or stems.

**Character Tokenizers**: These tokenizers split text into individual characters.

**Sentence Tokenizers**: These tokenizers break text into sentences.

**Whitespace Tokenizers**: These tokenizers split text based on whitespace characters.

**Punctuation-Based Tokenizers**:These tokenizers use punctuation marks as delimiters to split text.

**Rule-Based Tokenizers**: These tokenizers use predefined rules and patterns to split text.

**Pretrained Tokenizers**:Tokenizers that come as part of pretrained models and are designed to work with specific models.

**Hybrid Tokenizers**:Combine different tokenization strategies to address the complexities of various languages and tasks.






# Activity Objective:

 In this activity the textual data is preprocessed using **BERT tokenizer**. A BERT tokenizer is a specialized tokenizer used for preparing text input for BERT (Bidirectional Encoder Representations from Transformers), a popular deep learning model developed by Google for natural language processing (NLP). The BERT tokenizer is designed to handle the preprocessing steps required to convert raw text into a format that BERT can understand and work with effectively. This involves transforming a raw sentence into token embeddings, which can be used for various natural language processing (NLP) tasks such as sentiment analysis, text classification, named entity recognition, and more.

**Working of BERT Tokenizer**


1.   **Breaking Text into Pieces**: The BERT tokenizer starts by breaking down the input text into smaller parts called tokens. These tokens are not always full words; they can be pieces of words or even characters.
2.   **Adding Special Markers**: It adds special markers to help the model understand the structure of the text:


      *   [CLS]: This is added at the beginning of the text and helps the model get a summary of the whole input.
      *   [SEP]: This is used to separate different parts of the text, like separating a question from its context.

3. **Converting Tokens to Numbers**: Each token is turned into a number using a pre-defined list (vocabulary) that BERT knows about.
4. **Handling Padding**:Adding Padding tokens to make all the texts of the same length.
5. **Creating Attention Masks**: This helps the model know which tokens are real and which are just padding. It ensures the model focuses only on the actual content.
6. **Final Output**:  The final output includes a list of numbers representing the tokens, special markers, and an attention mask to guide the model.

```python
import warnings
warnings.filterwarnings('ignore')
#BertTokenziner and Bertmodel and torch are imported

from transformers import BertTokenizer, BertModel

import torch
```


In this activity a bert-base-uncased BERT Tokenizer model is used. The term "uncased" means that the model does not differentiate between uppercase and lowercase letters. BERT is designed to understand the context of words in a sentence by looking at both the words before and after them, rather than just one direction (left-to-right or right-to-left). This bidirectional approach helps it capture the meaning of words in a more nuanced way.

Tokenization involves splitting the sentence into subword units, while conversion maps these tokens to their unique numerical IDs in the BERT vocabulary.

This process prepares textual data for further processing by BERT models, essential for tasks like sentiment analysis or text classification. The code prints the tokens and their corresponding IDs, illustrating how BERT handles input text.



```python
# Load pre-trained BERT tokenizer and model

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

model = BertModel.from_pretrained('bert-base-uncased')

# Example sentence

sentence = "Emeritus Erulearning Solutions is a leading provider of online educational programs, specializing in delivering high-quality, flexible learning experiences for professionals seeking to enhance their skills and advance their careers. Leveraging partnerships with top universities and industry experts, Emeritus offers a range of executive education courses, certifications, and specialized training programs."
# Tokenize the sentence

tokens = tokenizer.tokenize(sentence)

print("Tokens:", tokens)

# Convert tokens to token IDs

token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Token IDs:", token_ids)
```

Once the token_ids are generated, Special tokens [CLS] and [SEP] are added, the tokens are converted into a tensor, and then the BERT model is used to generate embeddings.
The code then extracts the embeddings corresponding to the input tokens, excluding special tokens, for further use in natural language processing tasks.

## Add special tokens [CLS] and [SEP]
BERT requires special tokens for input sequences.[tokenizer.cls_token_id] is added at the start to aggregate information for classification tasks, and [tokenizer.sep_token_id] is added at the end to denote the end of the sequence. These tokens are crucial for BERT to understand the boundaries and roles of different parts of the input text.


```python
# Add special tokens [CLS] and [SEP]

token_ids = [tokenizer.cls_token_id] + token_ids + [tokenizer.sep_token_id]

print("Token IDs with Special Tokens:", token_ids)
```

## Convert Token IDs to Tensor:
The torch.tensor(token_ids).unsqueeze(0) is used to Convert the list of token IDs into a PyTorch tensor. unsqueeze(0) is used to add a batch dimension, setting the batch size to 1, which is required by the BERT model input format. This step prepares the data for processing by the model.


```python
# Convert token IDs to tensor

input_ids = torch.tensor(token_ids).unsqueeze(0)  # Batch size 1

print("Input IDs:", input_ids)


```

## Get the Embeddings from the BERT Model:

To get the embeddings from the BERT Model, torch.no_grad() is used. This context disables gradient computation, reducing memory consumption and speeding up computations.

```python
# Get the embeddings from the BERT model

with torch.no_grad():
    outputs = model(input_ids)
    embeddings = outputs.last_hidden_state



```

## Extract Embeddings for the Tokens:

For extracting the embeddings from the Token, embeddings[0][1:-1] is used. This extracts the embeddings for the actual tokens in the input sentence, excluding the embeddings for the special [CLS] and [SEP] tokens. This is done because we are interested in the embeddings for the original sentence tokens for further analysis or processing.

```python
# Extract embeddings for the tokens
token_embeddings = embeddings[0][1:-1]  # Exclude [CLS] and [SEP]

print("Token Embeddings Shape:", token_embeddings.shape)
print("Token Embeddings:", token_embeddings)
```
